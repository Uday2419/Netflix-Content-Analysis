```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../dataset/netflix_titles.csv")
```

```python
df.head()
```

```python
df.info
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.duplicated().sum()
```

```python
df["director"]=df["director"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["cast"]=df["cast"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["country"]=df["country"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["rating"] = df["rating"].fillna("Not Rated")
```

```python
df.isnull().sum()
```

```python
df["date_added"] = df["date_added"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["duration"]=df["duration"].fillna(df["duration"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df["date_added"]=pd.to_datetime(df["date_added"],errors="coerce")
```

```python
df.head()
```

```python
df["added_year"] = df["date_added"].dt.year
```

```python
df.head()
```

```python
df["added_month"]=df["date_added"].dt.month_name()
```

```python
df.head()
```

```python
df.to_csv("../dataset/netflix_cleaned.csv", index=False)
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="type")

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Count")

plt.tight_layout()
plt.savefig("../images/movies_vs_tvshows.png")

plt.show()
```

```python
country_counts = (
    df["country"]
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
    .reset_index()
)

country_counts.columns = ["Country", "Count"]

plt.figure(figsize=(10,5))

sns.barplot(data=country_counts, x="Country", y="Count")

plt.title("Top 10 Countries Producing Netflix Content")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("../images/top10_countries.png")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="rating",
    order=df["rating"].value_counts().index
)

plt.title("Distribution of Netflix Ratings")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("../images/ratings_distribution.png")

plt.show()
```

```python
added_year_counts = (
    df["added_year"]
    .value_counts()
    .sort_index()
    .reset_index()
)

added_year_counts.columns = ["Year", "Count"]

plt.figure(figsize=(10,5))

sns.lineplot(
    data=added_year_counts,
    x="Year",
    y="Count",
    marker="o"
)

plt.title("Content Added Over the Years")

plt.tight_layout()
plt.savefig("../images/content_added_over_years.png")

plt.show()
```

```python
genre_counts = (
    df["listed_in"]
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
    .reset_index()
)

genre_counts.columns = ["Genre", "Count"]

plt.figure(figsize=(10,5))

sns.barplot(data=genre_counts, x="Genre", y="Count")

plt.title("Top 10 Netflix Genres")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("../images/top10_genres.png")

plt.show()
```

```python
release_counts = (
    df["release_year"]
    .value_counts()
    .sort_index()
    .reset_index()
)

release_counts.columns = ["Year", "Count"]

plt.figure(figsize=(12,5))

sns.lineplot(
    data=release_counts,
    x="Year",
    y="Count"
)

plt.title("Netflix Content Release Trend")

plt.tight_layout()
plt.savefig("../images/release_trend.png")

plt.show()
```

```python
top_directors = (
    df[df["director"] != "Unknown"]["director"]
    .value_counts()
    .head(10)
    .reset_index()
)

top_directors.columns = ["Director", "Count"]

plt.figure(figsize=(12,5))

sns.barplot(data=top_directors, x="Director", y="Count")

plt.title("Top 10 Directors on Netflix")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.savefig("../images/top_directors.png")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=["int64", "float64"])

plt.figure(figsize=(8,5))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="Blues",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.savefig("../images/correlation_heatmap.png")

plt.show()
```

# Key Insights

- **Netflix has significantly more Movies than TV Shows.**
- **The United States contributes the highest amount of content.**
- **Most content was added between 2016 and 2019.**
- **TV-MA is one of the most common content ratings.**
- **International Movies and Dramas are among the most popular genres.**
- **Most movies have a duration between 90 and 120 minutes.**
